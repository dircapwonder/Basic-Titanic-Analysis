**1-Import Libraries & Dataset**

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("titanic.csv")
sns.set_style("whitegrid")
```

```python
print(df)
```

**2-Inspect the shape, columns, and basic information**

```python
print("Info:")
print(df.info())
print("-" * 30)
print("Shape:\n", df.shape)
print("-" * 30)
print("Columns:\n", df.columns)
```

**3-Check for missing values and handle them appropriately**

```python
print("Missing Values:")
df.isna().sum()
```

Handling Missing Values:

```python
df["Age"] = df["Age"].fillna(int(df["Age"].mean()))
df["Cabin"] = df["Cabin"].fillna("Unknown")
df["Embarked"] = df["Embarked"].fillna("Unknown")
df.isna().sum()
```

Output:

**4- Find how many passengers survived and did not survive**

```python
plt.figure(figsize=(5,3))
sns.countplot(x="Survived", data=df)
plt.title("Survival Count")
plt.show()
```

**5&6- Analyze survival based on gender & passenger class** 

```python
plt.figure(figsize=(10, 6))
sns.countplot(x="Survived", data=df, hue="Sex", palette="Set2")
plt.title("Survival Count by Gender")
plt.show()
```

```python
plt.figure(figsize=(10, 6))
sns.countplot(x="Pclass", data=df, hue="Survived", palette="Set2")
plt.title("Survival Count by Passenger Class")
plt.show()
```

**7-Find the average age**

```python
plt.figure(figsize=(8,6))
sns.histplot(df["Age"], kde=True, color="Brown")
plt.title("Average Age")
plt.xlabel("Age")
plt.ylabel("Count")
plt.show()
```

**Bonus 1: Fare distribution for survivors vs non-survivors**

```python
plt.figure(figsize=(8,6))
sns.histplot(df[df["Survived"]==1]["Fare"], label="Survived", color="Green")
sns.histplot(df[df["Survived"]==0]["Fare"], label="Deceased", color="Red")
plt.title("Average Fare")
plt.legend()
plt.xlabel("Fare")
plt.ylabel("Count")
plt.show()
```

**Bonus 2: Compare survival rates across age groups**

```python
df["AgeGroup"] = pd.cut(
    df["Age"], 
    bins=[0, 12, 18, 35, 60, 100], 
    labels=["Child", "Teen", "Young Adult", "Adult", "Elder"])

print("\nSurvival rate by age group:")
print(df.groupby("AgeGroup")["Survived"].mean())

plt.figure(figsize=(6,4))
sns.barplot(x="AgeGroup", y="Survived", data=df, color="Brown")
plt.title("Average Survival Rate - Age Group")
plt.show()
```

**Bonus 3: Survival Rate Across Gender**

```python
print("\nSurvival rate by gender:")
print(df.groupby("Sex")["Survived"].mean())

plt.figure(figsize=(6,4))
sns.barplot(x="Sex", y="Survived", data=df, color="Brown")
plt.title("Average Survival Rate - Gender")
plt.show()
```
